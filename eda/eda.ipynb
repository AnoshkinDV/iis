```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set_theme(style="whitegrid")

os.makedirs("eda", exist_ok=True)
os.makedirs("data", exist_ok=True)
```

```python
DATA_PATH = "../data/train.csv"
```

## Загрузка и описание данных

```python
df = pd.read_csv(DATA_PATH)

print("Размер датасета:", df.shape)
print("\nПервые 5 строк:")
display(df.head())

print("\nСписок столбцов:")
print(df.columns.tolist())
```

```python
print("Информация о датасете:")
df.info()

print("\nОписательная статистика:")
display(df.describe())

print("\nТипы данных:")
display(df.dtypes)
```

```python
print("Количество пропусков по столбцам:")
display(df.isna().sum())

print("\nКоличество дубликатов:", df.duplicated().sum())
```

## Числовые, категориальная и целевая переменные

```python
target_col = "price_range"

categorical_cols = ["blue", "dual_sim", "four_g", "three_g", "touch_screen", "wifi"]

numeric_cols = [col for col in df.columns if col not in categorical_cols + [target_col]]

print("Целевая переменная:", target_col)
print("\nКатегориальные признаки:")
print(categorical_cols)

print("\nЧисловые признаки:")
print(numeric_cols)
```

## Очистка данных

```python
df_clean = df.copy()

# Удаление дубликатов
df_clean = df_clean.drop_duplicates()

# Если вдруг будут пропуски в числовых — заполняем медианой
for col in numeric_cols:
    if df_clean[col].isna().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Если вдруг будут пропуски в категориальных — заполняем модой
for col in categorical_cols:
    if df_clean[col].isna().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Размер после очистки:", df_clean.shape)
print("Пропуски после очистки:")
display(df_clean.isna().sum())
```

## Визуализация

1. График распределения целевой переменной

```python
plt.figure(figsize=(8, 5))
sns.countplot(x=target_col, data=df_clean)
plt.title("Распределение целевой переменной price_range")
plt.xlabel("Класс price_range")
plt.ylabel("Количество объектов")
plt.tight_layout()
plt.savefig("eda/graph1_price_range_distribution.png")
plt.show()
```

**Вывод:** распределение целевой переменной `price_range` близко к равномерному — 
каждый класс содержит примерно одинаковое количество наблюдений. 
Это положительно влияет на обучение моделей классификации, поскольку отсутствует дисбаланс классов, 
который мог бы привести к ухудшению качества предсказаний для редких классов.

2. График распределения RAM

```python
plt.figure(figsize=(8, 5))
sns.histplot(df_clean["ram"], bins=30, kde=True)
plt.title("Распределение признака RAM")
plt.xlabel("RAM")
plt.ylabel("Частота")
plt.tight_layout()
plt.savefig("eda/graph2_ram_distribution.png")
plt.show()
```

**Вывод:** распределение признака `ram` охватывает широкий диапазон значений и не имеет выраженной асимметрии. 
Это говорит о том, что признак представлен в данных достаточно полно и может играть важную роль при определении диапазона цены устройства (`price_range`). 
В дальнейшем стоит отдельно проверить зависимость между `ram` и целевой переменной.

3. График зависимости RAM от price_range

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="price_range", y="ram", data=df_clean)
plt.title("Зависимость RAM от price_range")
plt.xlabel("Класс price_range")
plt.ylabel("RAM")
plt.tight_layout()
plt.savefig("eda/graph3_ram_vs_price_range.png")
plt.show()
```

**Вывод:** график показывает сильную положительную зависимость между объемом оперативной памяти (`ram`) и целевой переменной `price_range`. 
С увеличением объема RAM медианное значение для классов `price_range` также увеличивается. 
Это свидетельствует о том, что устройства с большим объемом оперативной памяти относятся к более высоким ценовым категориям. 

Таким образом, признак `ram` может являться одним из ключевых факторов при построении модели для предсказания ценового диапазона мобильного устройства.


4. Корелляционная матрица

```python
plt.figure(figsize=(14, 10))
corr = df_clean.corr(numeric_only=True)
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Корреляционная матрица числовых признаков")
plt.tight_layout()
plt.savefig("eda/graph4_correlation_matrix.png")
plt.show()
```

**Вывод:** корреляционная матрица показывает степень взаимосвязи между числовыми признаками датасета. 
Большинство признаков имеют слабую корреляцию друг с другом, что говорит об отсутствии сильной мультиколлинеарности между переменными.

Наиболее заметная положительная корреляция наблюдается между признаками `ram` и целевой переменной `price_range`. 
Это подтверждает выводы предыдущих графиков о том, что объем оперативной памяти является одним из наиболее важных факторов, влияющих на ценовой диапазон мобильного устройства.

Также можно заметить умеренную корреляцию между некоторыми признаками экрана, например `px_height` и `px_width`, 
что логично, так как оба признака характеризуют разрешение дисплея.

В целом данные выглядят достаточно независимыми, что является положительным фактором для построения модели машинного обучения.

5. График scatter plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x="battery_power", y="ram", hue="price_range", alpha=0.7)
plt.title("Связь battery_power и ram с учетом price_range")
plt.xlabel("Battery Power")
plt.ylabel("RAM")
plt.tight_layout()
plt.savefig("eda/graph5_battery_ram_scatter.png")
plt.show()
```

**Вывод:** на графике можно наблюдать взаимосвязь между признаками `battery_power` и `ram` с учетом целевой переменной `price_range`. 
Наиболее заметно, что значения `ram` существенно влияют на класс `price_range`: устройства с большим объемом оперативной памяти чаще относятся к более высоким ценовым категориям.

При этом признак `battery_power` распределен более равномерно и не показывает такой явной зависимости от `price_range`, как `ram`. 
Однако комбинация признаков `battery_power` и `ram` позволяет лучше разделять объекты разных классов.

Таким образом, признак `ram` является одним из наиболее информативных для предсказания `price_range`, 
а `battery_power` может выступать дополнительным фактором, который в сочетании с другими характеристиками помогает модели точнее определять ценовой диапазон устройства.

6. Интерактивный график

```python
fig = px.scatter(
    df_clean,
    x="ram",
    y="battery_power",
    color="price_range",
    title="Интерактивный график: RAM vs Battery Power",
    hover_data=df_clean.columns,
)

fig.show()
fig.write_html("eda/graph6_interactive_ram_battery.html")
```
